```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('MultipleLR.csv - MultipleLR.csv (1).csv',header=None)
df.columns = ['x','y','z','label']
df
```

```python
X = df[['x','y','z']].values
y = df['label'].values
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
X = X/np.max(X,axis=0)
y = y/np.max(y)
X = np.c_[np.ones(X.shape[0]),X]

print(X)
print(y)
```

```python
n_features = X.shape[1]
weights = np.zeros(n_features)

lr = 0.01
epochs = 1000
```

```python
for epoch in range(epochs):
    for i in range(len(X)):
        xi = X[i]
        yi = y[i]

        y_pred = np.dot(xi,weights)
        error = y_pred-yi

        weights -= lr*error*xi
print('Trained weights: ',weights)
```

```python
y_pred = np.dot(X,weights)
mse = np.mean((y-y_pred)**2)
print('MSE: ',mse)

print('Actual: ',y[0:5])
print('Predicted: ',y_pred[0:5])
```
